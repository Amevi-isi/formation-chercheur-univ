## 7. Exemple de graphique simple

De nombreux notebooks du cours créent des graphiques. Exécutez la cellule suivante pour vérifier que l'affichage des graphiques fonctionne.

````python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [1, 4, 9, 16, 25]

plt.figure(figsize=(5, 3))
plt.plot(x, y, marker="o")
plt.title("Un graphique simple")
plt.xlabel("x")
plt.ylabel("y")
plt.grid(True, alpha=0.3)
plt.show()
````

## 8. Le noyau (kernel)

Le **noyau** est le processus Python qui exécute votre code. Il garde en mémoire les variables créées par les cellules déjà exécutées.

Actions utiles :

- **Interrompre** : arrêter une cellule qui tourne trop longtemps (bouton ⏹ ; sur Colab : **Exécution → Interrompre l'exécution**).
- **Redémarrer** : réinitialiser la mémoire et les variables (**Restart** dans VS Code ; **Exécution → Redémarrer la session** sur Colab).

Si un notebook se comporte bizarrement, la première solution est généralement :

1. Enregistrer le notebook.
2. Redémarrer le noyau.
3. Réexécuter les cellules depuis le haut.

## 9. Problèmes courants

**`ModuleNotFoundError` (bibliothèque introuvable)**

- Sur Colab : exécutez d'abord la cellule **⚙️ Configuration** du notebook.
- Dans VS Code : vérifiez que le noyau sélectionné est bien `.venv` (en haut à droite). Sinon, réinstallez avec `pip install -r requirements.txt` dans le terminal, environnement activé.

**`No module named 'plotting_utils'`**

- Sur Colab : exécutez la cellule **⚙️ Configuration**.
- Dans VS Code : ouvrez le notebook depuis le dossier `Labs/` du dépôt, pas une copie isolée.

**Une variable n'est pas définie**

Réexécutez les cellules précédentes depuis le haut. Les notebooks dépendent de l'ordre d'exécution.

**Les curseurs interactifs ne s'affichent pas**

Réexécutez la cellule. Dans VS Code, vérifiez que l'extension Jupyter est à jour.

**`CERTIFICATE_VERIFY_FAILED` lors d'un téléchargement (macOS)**

Avec le Python téléchargé sur python.org, exécutez une fois **Applications → Python 3.x → Install Certificates.command**, puis redémarrez le noyau.

**La session Colab s'est déconnectée**

Reconnectez-vous, puis réexécutez les cellules depuis le haut, à commencer par la configuration.
